final cheeck

Deadline passed: Expected failure
Because it is something that can normally happen.

Expected failure: return a result
Fault: exception
Because one is expected and the other is unexpected.

The most dangerous thing is returning "Success" when the operation actually failed.

User message: simple and clear.
Log: technical details for developers like timestamp, request ID, and stack trace.
